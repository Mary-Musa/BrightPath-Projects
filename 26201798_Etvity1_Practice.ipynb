## Task 8: Handle missing numerical values

### Question
The lecture introduced mean and median imputation.

1. Calculate the mean and median of `study_hours`.
2. Calculate the mean and median of `attendance`.
3. Fill missing `study_hours` using the median.
4. Fill missing `attendance` using the median.
5. Fill the invalid age that was converted to `NaN` using the median age.
6. Verify that no missing values remain in these three numerical columns.

### Challenge
Why might median imputation be preferable when the variable contains outliers?

```python
# Write your own code

print("Study hours - mean:", df_clean["study_hours"].mean())

print("Study hours - median:", df_clean["study_hours"].median())

print("Attendance - mean:", df_clean["attendance"].mean())

print("Attendance - median:", df_clean["attendance"].median())

#filling  study_hours with median
study_median = df_clean["study_hours"].median()
df_clean["study_hours"] = df_clean["study_hours"].fillna(study_median)

# filling attendance with median
attendance_median = df_clean["attendance"].median()

df_clean["attendance"] = df_clean["attendance"].fillna(attendance_median)

#filling invalid age (now NaN) with median age
age_median = df_clean["age"].median()

df_clean["age"] = df_clean["age"].fillna(age_median)

print("\n6. Verification - remaining missing values in these columns:")

print(df_clean[["study_hours", "attendance", "age"]].isnull().sum())
```

Challenge: Because outliers do not affect the median as much as they do to the mean. That makes the median better for representation when the data contains the outliers.

## Task 9: Descriptive statistics from scratch

### Question
For `study_hours`, calculate:

1. Mean
2. Median
3. Mode
4. Minimum
5. Maximum
6. Range
7. Variance
8. Standard deviation

Do this using pandas methods.

Then verify the range manually using `max - min`.

**Discussion:** Which of these measures describes central tendency, and which describes spread?

```python
study = df_clean["study_hours"]

# Write your own code

study_mean     = study.mean()
study_median   = study.median()
study_mode     = study.mode()[0]
study_min      = study.min()
study_max      = study.max()
study_range    = study_max - study_min
study_variance = study.var()
study_std      = study.std()

print("Mean:", round(study_mean, 3))
print("Median:", study_median)
print("Mode:", study_mode)
print("Minimum:", study_min)
print("Maximum:", study_max)
print("Range:", study_range)
print("Variance:", round(study_variance, 3))
print("Standard deviation:", round(study_std, 3))
print("Manual range check:", study.max() - study.min())
```

**Discussion**:The mean, median and mode describe the central tendency while the range,variance and standard deviation describe the spread.

### Question
Create a new numerical variable called `attendance_with_outlier` by adding an extreme value of 5 to the attendance data.

Calculate the mean and median before and after adding the extreme value.

What do you observe?

### Question

Answer briefly:

1. Which stage of the Data Science Lifecycle did you find most important and why?
2. Why is data cleaning performed before modelling?
3. Why can a visually obvious relationship still be misleading?
4. Why should the target variable be separated from the input features?
5. What additional data would you collect to improve a student-performance analysis?

### Solution



1. I found data cleaning the most important stage because if the data is not correct, the results of the analysis may also be wrong. Cleaning the data helps remove things like missing values, duplicates, and errors.
