```python
# a) Write a regular expression for validating UL student email addresses in this format: studentID@studentmail.ul.ie	,where studentID is an 8-digit number

import re   # import re library for regex


def StudentEmailValidator(email):


    #Python String Formatting https://www.w3schools.com/python/python_string_formatting.asp

    print (f'\n email = {email}')

    #Python RegEx https://www.w3schools.com/python/python_regex.asp

    match = re.search(r"^[0-9]{8}@studentmail\.ul\.ie$", email)
    #YOUR CODE HERE

    if match:
      print("Valid Student email")

    else:
        print("Invalid Student email")


StudentEmailValidator("12345678@studentmail.ul.ie")
StudentEmailValidator("     12345678@studentmail.ul.ie")
StudentEmailValidator("12345678@studentmail.ul.ie       ")
StudentEmailValidator("123456789@studentmail.ul.ie")
StudentEmailValidator("1234567@studentmail.ul.ie")
```

```python
# b) Write a regular expression for validating UL staff email addresses in this format:
# FirstName.MiddleNameInitial.LastName@ul.ie ,where FirstName and lastName each consist of 2 to 20 Latin letters, and MiddleNameInitial (a single letter) is optional.

import re   # import re library for regex


def staffEmailValidator(email):
    email = email.strip(); #remove Leading & Trailing Spaces
    print (f'\n email = {email}')

    match = re.search(r"^[A-Za-z]{2,20}(?:\.[A-Za-z])?\.[A-Za-z]{2,20}@ul\.ie$", email)
    if not match:
        print('Invalid UL Staff Email pattern')
    else:
        print('Valid UL Staff Email pattern')


staffEmailValidator("arash@ul.ie")
staffEmailValidator("Arash.joorabchi@ul.ie")
staffEmailValidator("Arash.A.joorabchi@ul.ie") # Write more tests to check all the edge cases.
```
